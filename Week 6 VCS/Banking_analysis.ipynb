# 4. Filtering with boolean indexing

```python
# All ATM cash withdrawals
atm_withdrawals=df[(df["channel"]=="ATM")&(df["transaction_type"]=="Cash Withdrawal")]
print(f"ATM cash withdrawals:{len(atm_withdrawals)}")
atm_withdrawals.head()
```

```python
# All failed or reversed transactions
not_successful=df[df["transaction_status"]!="Success"]
print(f"Not successful:{len(not_successful)} out of {len(df)} ({len(not_successful)/len(df):.1%})")
not_successful["transaction_status"].value_counts()
```

```python
# High-value fund transfers (Nepal context: NRB requires extra scrutiny above certain thresholds)
large_transfers=df[(df["transaction_type"]=="Fund Transfer")&(df["amount_npr"]>50000)]
print(f"Large fun transfers (> NPR 50,000): {len(large_transfers)}")
```

# 5. Sorting

```python
# Top 10 highest-value transactions overall
df.sort_values("amount_npr",ascending=False).head(10)[
    ["transaction_id","branch_name","transaction_type","amount_npr","transaction_status"]
]
```

# Line Chart : Monthly Transaction Volume Over 2024

```python
monthly = df.groupby("year_month").size()

plt.figure()

plt.plot(
    monthly.index,
    monthly.values,
    marker="o",
    color="#2E7D32"
)

plt.title("Monthly Transaction Volume, 2024")
plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)

plt.axvspan(
    "2024-10",
    "2024-11",
    color="orange",
    alpha=0.15,
    label="Dashain/Tihar season"
)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
channel_counts = df["channel"].value_counts()

plt.figure()

sns.barplot(
    x=channel_counts.index,
    y=channel_counts.values,
    hue=channel_counts.index,
    palette="viridis",
    legend=False
)

plt.title("Transaction Count by Channel")
plt.xlabel("Channel")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# Boxplot

```python
plt.figure()

sns.boxplot(
    data=df,
    x="channel",
    y="processing_time_ms",
    hue="channel",
    palette="Set2",
    legend=False
)

plt.title("Processing Time by Channel")
plt.xlabel("Channel")
plt.ylabel("Processing Time (ms)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# Heatmap: Channel vs transaction status

```python
pivot = pd.crosstab(df["channel"], df["transaction_status"])

plt.figure()
sns.heatmap(
    pivot,
    cmap="YlGnBu",
    annot=True,
    fmt="d",
)
plt.title("Transaction Status by Channel")
plt.xlabel("Status")
plt.ylabel("Channel")
plt.tight_layout()
plt.show()
```

```python
account_counts = df["account_type"].value_counts()

plt.figure()
plt.pie(account_counts.values, labels=account_counts.index, autopct="%1.1f%%",
        colors=sns.color_palette("pastel"))
plt.title("Transaction share by Account Type")
plt.tight_layout()
plt.show()
```
